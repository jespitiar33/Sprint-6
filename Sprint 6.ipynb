## Plan de acción para procesamiento ##

# Prepara los datos

```python
# Reemplazamos los nombres de las columnas (ponlos en minúsculas).
# Ponemos los nombres de las columnas en snake_case

df_games.columns = df_games.columns.str.lower().str.replace(' ', '_')
```

```python
df_games.head()
```

# Columna Names

```python
# Daremos formato a la columna Names para que tenga una mejor visualización.
# Convertiremos a mayúsculas
df_games['name'] = df_games['name'].str.upper()
# Quitar espacios no deseados
df_games['name'] = df_games['name'].str.strip()
# Quitar signos de pausa
df_games['name'] = df_games['name'].str.replace('.','').str.replace(',','').str.replace(';','').str.replace(':','')

df_games.sample()
```

```python
# Identicar los casos perdidos
df_games[df_games['name'].isna()]
```

```python
# Solo tenemos dos casos de juegos con datos perdidos en el nombre, no hay forma de saber cuales son.

# Puede ser que estos datos estén ausentes por un error en el procesamiento del archivo, vemos que son de 1993 por lo que el ser más antiguos estos datos son más suceptibles perdida o daño.

# Reemplazaremos estos dos casos por el nombre "Not Defined".

df_games.loc[df_games['name'].isna(),'name'] = 'NOT DEFINED'
```

# Columna Platform

```python
# Define una función para asignar el nombre del fabricante de la plataforma para poder agrupar de ese modo posteriormente y facilitar el analisis.

def GrupoPlat (x) :
    if x in ['3DS','DS','GC','N64','NES','SNES','Wii','WiiU']:
        return 'Nintendo'
    elif x in ['PC','PCFX']:
        return 'PC'
    elif x in ['PS','PS2','PS3','PS4','PSP','PSV']:
        return 'PlayStation'
    elif x in ['X360','XB','XOne']:
        return 'Xbox'
    else:
        return 'Other'
```

```python
# Creamos una columna con la información anterior.

df_games['platform_group'] = df_games['platform'].apply(GrupoPlat)
df_games.sample(5)
```

```python
# Ahora podemos visualizar los juegos por fabricante de consolas.

df_games['platform_group'].value_counts()
```

# Columna Year of Release

```python
#Eliminamos valores perdidos, no es práctico investigar el año de lanzamiento de todos los juegos a los que les falta este dato.

df_games = df_games.dropna(subset = ['year_of_release'])
df_games.info()
```

```python
#Transformamos a entero el año ya que no es correcto que este dato esté como float.
df_games['year_of_release'] = df_games['year_of_release'].astype(int)
df_games.info()
```

# Columna genre

```python
# Identificamos los valores perdidos 
df_games[df_games['genre'].isna()]
```

```python
# Solo tenemos dos valores perdidos que son los mismos juegos que no tienen nombre por lo que también los indicatemos como Not Defined para que sea más práctico el estudio de la tabla.
df_games['genre'] = df_games['genre'].fillna('Not Defined')
```

# Columna user_score

```python
# Corregimos el caso del dato tbd y lo cambiamos a nan para poder seguir manipulando la tabla
df_games.loc[df_games['user_score'] == 'tbd', 'user_score'] = np.nan

# Transformar el tipo de variable a float

df_games['user_score'] = df_games['user_score'].astype(float)

# Estadarizar con critic_score para que tengan el mismo formato sobre 100 (multiplicando por 10).

df_games['user_score'] = df_games['user_score']*10

print(df_games['user_score'].describe())

print(df_games.sample(2))
```

```python
# Verificamos el cambio

df_games['user_score'].unique()
```

# Columna rating

```python
# Crear función para tener nombre del dato rating para en vez de tener solo las siglas tener también la palabra completa que describe el dato

def Rat_Sem (x):
    if x == 'E' or x == 'K-A':
        return 'Everyone'
    if x == 'T':
        return 'Teens'
    if x == 'E10':
        return 'Everyone 10y'
    if x == 'EC':
        return 'Early childhood'
    if x == 'RP':
        return 'Pending'
    if x == 'M':
        return 'Mature'
    if x == 'AO':
        return 'Adults Only'
    else:
        return 'Pending'
```

```python
# Creamos la columna rating_sem
df_games['rating_sem'] = df_games['rating'].apply(Rat_Sem)

df_games.sample(5)
```

```python
# Visualizamos cuantos juegos están en cada categoría

df_games['rating_sem'].value_counts()
```

# Columna total_sales

```python
# Creamos una columna con el total global de ventas.

df_games['total_sales'] = df_games[['na_sales','eu_sales','jp_sales','other_sales']].sum(axis=1)
df_games.sample(10)
```

```python
# Creamos una función que itinire sobre la columna year_of_release y elimine la información de las filas que sean de 1994 hacia atras.

def remove_old (df_main_platforms):
  for year in df_main_platforms['year_of_release']:
    if year <= 2011:
      df_main_platforms = df_main_platforms.drop(df_main_platforms[df_main_platforms['year_of_release'] == year].index)
  return df_main_platforms
```

# 5 Prueba la siguientes hipótesis

— Las calificaciones promedio de los usuarios para las plataformas Xbox One y PC son las mismas.

```python
# Creamos un data frame con nombre df_games_hipotesis con las columnas name, platform, user_score, rating_sem,total_sales.

df_games_hipotesis = df_games[['name','platform','user_score','genre','total_sales']]
print(df_games_hipotesis.head(10))
```

```python
# Eliminamos las líneas con valores nulos en la columna user_score

df_games_hipotesis = df_games_hipotesis.dropna(subset=['user_score'])
print(df_games_hipotesis.head(10))
```

```python
# Sacamos una muestra de 100 scores de cada columna de user_score para XOne y para PC

df_xbox_sample = df_games_hipotesis[df_games_hipotesis['platform'] == 'XOne'].sample(100)['user_score'].reset_index(drop=True)
df_xbox_pc_sample = df_games_hipotesis[df_games_hipotesis['platform'] == 'PC'].sample(100)['user_score'].reset_index(drop=True)
print(df_xbox_sample.head(10))
print(df_xbox_pc_sample.head(10))
```

```python
# Usamos la hipótesis sobre la igualdad de las medias de dos poblaciones

# Usamos la hipótesis nula de que las medias de dos poblaciones estadísticas son iguales en las plataformas Xbox One y PC, se acepta o rechaza la hipotesis nula. Valor p=5%
# La hipotesis alternativa es que las calificaciones promedio de los usuarios para las plataformas Xbox One y PC son diferentes.

alpha = 0.05
results = st.ttest_ind(df_xbox_sample, df_xbox_pc_sample)

print('valor p: ', results.pvalue)
if results.pvalue < alpha:
    print('No podemos rechazar la hipótesis nula, las calificaciones promedio son las mismas')
else:
    print('Rechazamos la hipótesis nula, las calificaciones promedio no son las mismas')
```

— Las calificaciones promedio de los usuarios para los géneros de Acción y Deportes son diferentes.

```python
# hipótesis nula: Las calificaciones promedio de los usuarios para los géneros de Acción y Deportes son diferentes.
# Sacamos una muestra de 100 scores de cada columna de user_score para XOne y para PC

df_xbox_sample_genre = df_games_hipotesis[df_games_hipotesis['genre'] == 'Action'].sample(100)['user_score'].reset_index(drop=True)
df_xbox_pc_sample_genre = df_games_hipotesis[df_games_hipotesis['genre'] == 'Sports'].sample(100)['user_score'].reset_index(drop=True)
print(df_xbox_sample_genre.head(10))
print(df_xbox_pc_sample_genre.head(10))
```

```python
# Usamos la hipótesis sobre la igualdad de las medias de dos poblaciones

# Usamos la hipótesis nula de que las medias de dos poblaciones estadísticas son diferentes en los generos de acción y deportes, Valor p=5%
# La hipotesis alternativa es que las calificaciones promedio de los usuarios para los géneros de Acción y Deportes son iguales.
alpha = 0.05
results = st.ttest_ind(df_xbox_sample_genre, df_xbox_pc_sample_genre)

print('valor p: ', results.pvalue)
if results.pvalue < alpha:
    print('No podemos rechazamos la hipótesis nula, el promedio de las calificaciones de los usuarios para los géneros de Acción y Deportes son diferentes')
else:
    print('Rechazamos la hipótesis nula, el promedio de las calificaciones de los usuarios para los géneros de Acción y Deportes no son diferentes')
```
